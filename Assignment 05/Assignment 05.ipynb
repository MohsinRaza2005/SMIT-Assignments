```python
"""Simple Shopping Cart: Product + ShoppingCart classes with a CLI menu."""


class Product:
    def __init__(self, name, price, quantity=1):
        if not name.strip():
            raise ValueError("Name cannot be empty.")
        if price < 0:
            raise ValueError("Price cannot be negative.")
        if quantity <= 0:
            raise ValueError("Quantity must be at least 1.")
        self.name, self.price, self.quantity = name.strip(), price, quantity

    def total(self):
        return self.price * self.quantity

    def display(self):
        print(f"  {self.name:<15} ${self.price:>7.2f} x {self.quantity:<3} = ${self.total():>8.2f}")


class ShoppingCart:
    def __init__(self):
        self._items = {}  # lowercase name -> Product

    def add_product(self, product):
        key = product.name.lower()
        if key in self._items:  # same product: just increase quantity
            self._items[key].quantity += product.quantity
        else:
            self._items[key] = product

    def remove_product(self, name, quantity=None):
        item = self._items.get(name.strip().lower())
        if item is None:
            raise ValueError(f"'{name}' is not in your cart.")
        if quantity is None or quantity == item.quantity:
            del self._items[item.name.lower()]
        elif quantity <= 0:
            raise ValueError("Quantity to remove must be at least 1.")
        elif quantity > item.quantity:
            raise ValueError(f"You only have {item.quantity} of '{item.name}'.")
        else:
            item.quantity -= quantity

    def calculate_total(self):
        return sum(item.total() for item in self._items.values())

    def display_cart(self):
        if not self._items:
            print("\nYour cart is empty.")
            return
        print("\n------------- YOUR CART -------------")
        for item in self._items.values():
            item.display()
        print("-------------------------------------")

    def show_total(self):
        print(f"Total: ${self.calculate_total():.2f}")

    def clear(self):
        self._items.clear()

    def is_empty(self):
        return not self._items


def ask(prompt, kind=float):
    """Keep asking until the user enters a valid number."""
    while True:
        try:
            return kind(input(prompt))
        except ValueError:
            print("  Please enter a valid number.")


def main():
    cart = ShoppingCart()
    while True:
        print("\n1. Add  2. View  3. Remove  4. Clear  5. Check out  6. Quit")
        choice = input("Choose an option: ").strip()
        try:
            if choice == "1":
                name = input("Product name: ")
                price = ask("Price: $")
                cart.add_product(Product(name, price, ask("Quantity: ", int)))
                print("Added to cart.")
            elif choice == "2":
                cart.display_cart()
                if not cart.is_empty():
                    cart.show_total()
            elif choice == "3":
                name = input("Product to remove: ")
                amount = input("Quantity (Enter = all): ").strip()
                cart.remove_product(name, int(amount) if amount else None)
                print("Cart updated.")
            elif choice == "4":
                cart.clear()
                print("Cart cleared.")
            elif choice == "5":
                if cart.is_empty():
                    print("Your cart is empty.")
                else:
                    cart.display_cart()
                    cart.show_total()
                    print("Thank you for your purchase!")
                    cart.clear()
            elif choice == "6":
                print("Goodbye!")
                break
            else:
                print("Invalid choice, enter 1-6.")
        except ValueError as error:
            print(f"Error: {error}")


if __name__ == "__main__":
    main()
```
