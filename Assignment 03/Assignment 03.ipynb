# Question#13

```python
menu = []

for i in range(5):
    dish = input("Enter a dish name: ")

    if dish.lower() == "fish":
        print("Fish is not served. Skipping...")
        continue

    menu.append(dish)

print("Restaurant Menu:")
print(menu)
```

# Question#14

```python
discard_dishes =["fish","nihari","qeema"]
menu = []
count = 0 
while True:
    dish = input("Enter your dish name:").lower()
    
    if dish in discard_dishes:
        print("This is discard dish")
        continue
    menu.append(dish)
    count+=1 
    print(f"{count}: {dish} added in menu")

    if count == 5:
        break 

print(menu)
    
```
