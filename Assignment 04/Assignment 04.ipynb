# Assignment 04: Library Management System



# Step 1: Lists to store data

```python
# one book = one dictionary, all books = one list
books = []

# every time a book is issued, its ID is added here
# (this is how we track loans)
loans = []
```

# Step 2: Helper functions

```python
# search the list with a for loop
def find_book(book_id):
    for book in books:
        if book["id"] == book_id:
            return book
    return None


# print the details of one book
def show_book(book):
    print("Book ID          :", book["id"])
    print("Title            :", book["title"])
    print("Author           :", book["author"])
    print("Total stock      :", book["total"])
    print("Available copies :", book["available"])
```

# Feature 1: Add Book

```python
def add_book():
    book_id = int(input("Enter book ID: "))

    # Challenge: no duplicate IDs
    if find_book(book_id) != None:
        print("This book ID already exists.")
        return

    title = input("Enter title: ")
    author = input("Enter author: ")
    copies = int(input("Enter number of copies: "))

    # create a dictionary for the new book
    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "total": copies,
        "available": copies
    }

    books.append(book)
    print("Book added successfully!")
```

# Feature 2: View All Books

```python
def view_books():
    if len(books) == 0:
        print("No books available.")
    else:
        for book in books:
            show_book(book)
            print("-------------------------")
```

# Feature 3: Search Book

```python
def search_book():
    book_id = int(input("Enter book ID to search: "))

    for book in books:
        if book["id"] == book_id:
            show_book(book)
            return

    print("Book not found.")
```

# Feature 4: Update Book

```python
def update_book():
    book_id = int(input("Enter book ID to update: "))
    book = find_book(book_id)

    if book == None:
        print("Book not found.")
        return

    print("Press Enter if you do not want to change a value.")
    new_title = input("New title: ")
    new_author = input("New author: ")
    new_copies = input("New total copies: ")

    if new_title != "":
        book["title"] = new_title

    if new_author != "":
        book["author"] = new_author

    if new_copies != "":
        new_copies = int(new_copies)
        issued = loans.count(book_id)   # copies currently issued

        # total can not be less than the copies already issued
        if new_copies < issued:
            print("Copies can not be less than issued copies:", issued)
        else:
            book["total"] = new_copies
            book["available"] = new_copies - issued

    print("Book updated successfully!")
```

# Feature 5: Delete Book

```python
def delete_book():
    book_id = int(input("Enter book ID to delete: "))
    book = find_book(book_id)

    if book == None:
        print("Book not found.")
    elif book_id in loans:
        print("Can not delete. Some copies are still issued.")
    else:
        books.remove(book)
        print("Book deleted successfully!")
```

# Feature 6: Issue Book

```python
def issue_book():
    book_id = int(input("Enter book ID to issue: "))
    book = find_book(book_id)

    if book == None:
        print("Book not found.")
    elif book["available"] > 0:
        book["available"] -= 1      # decrease copies by one
        loans.append(book_id)       # record the loan
        print("Book issued successfully!")
    else:
        print("Book is not available.")
```

# Feature 7: Return Book

```python
def return_book():
    book_id = int(input("Enter book ID to return: "))
    book = find_book(book_id)

    if book == None:
        print("Book not found.")
    elif book_id not in loans:
        # Challenge: stock can not go above total copies
        print("This book was not issued. Nothing to return.")
    else:
        loans.remove(book_id)       # remove one loan record
        book["available"] += 1      # increase copies by one
        print("Book returned successfully!")
```

# Step 3: Main Menu

Run this cell to start the program.

```python
while True:
    print()
    print("LIBRARY MANAGEMENT SYSTEM")
    print("1. Add Book")
    print("2. View All Books")
    print("3. Search Book")
    print("4. Update Book")
    print("5. Delete Book")
    print("6. Issue Book")
    print("7. Return Book")
    print("8. Exit")

    choice = input("Enter your choice (1-8): ")

    if choice == "1":
        add_book()
    elif choice == "2":
        view_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        update_book()
    elif choice == "5":
        delete_book()
    elif choice == "6":
        issue_book()
    elif choice == "7":
        return_book()
    elif choice == "8":
        print("Goodbye!")
        break
    else:
        print("Invalid choice. Enter a number from 1 to 8.")
```
